```python
def longest_word(filename):
    with open(filename, 'r', encoding='utf-8') as f:
        text = f.read()

    text = text.replace('\n', ' ')

    for ch in ',.!?:;—-':
        text = text.replace(ch, '')

    words = text.split()
    return max(words, key=len)
```
